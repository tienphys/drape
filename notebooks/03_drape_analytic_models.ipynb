# Analytic DRAPE models

Three equivalent gate/circuit-level descriptions of the DRAPE protocol, in increasing order of analytic closure:

1. **Unitary form** — exponentiate everything with `qutip.expm`.
2. **Explicit matrix** — hand-derived 3 by 3 matrices for each gate.
3. **Closed form** — single expression for $p_1, p_2$.

All three live in the 3-level Hilbert space `{|0⟩, |1⟩, |2⟩}` and act on the `{|1⟩, |2⟩}` subspace.

## Sanity check: the four forms agree

For the closed form, we use the parametric signature with `beta0=0, xi=1, beta=-delta` so the internal `delta = -xi·(beta - beta0)` recovers the input `delta`.

```python
eps   = 0.10
delta = 0.01
ks    = dl.DEFAULT_REP_RANGE

print(f'{"k":>4} {"unitary":>12} {"matrix":>12} {"closed":>12} {"linear":>12}')
for idx_k, k in enumerate(ks):
    _, p_u = dl.DRAPE_unitary_probs(eps, delta, k)
    _, p_m = dl.DRAPE_matrix_probs (eps, delta, k)
    _, p_c = dl.DRAPE_closed_form(eps, delta, k)
    _, p_l = dl.DRAPE_closed_form_linear(eps, delta, k)
    print(f'{k:>4} {p_u:>12.6f} {p_m:>12.6f} {(1-p_c):>12.6f} {1-p_l:>12.6f}')
```
